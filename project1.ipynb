5.Handle missing values

```python
df["age"]=df["age"].fillna(df["age"].median())
```

```python
df[df.isnull().any(axis=1)]
```

```python
df[df["age"]==df["age"].median()]
```

```python
df["sex"]=df["sex"].fillna(df["sex"].mode()[0])
```

```python
df["localization"]=df["localization"].fillna(df["localization"].mode()[0])
```

```python
age=df["age"].to_numpy()
```
